```python
# importing the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Setup and Data Loading
sns.set(style="whitegrid")
df = pd.read_excel('cars24-car-price.xlsx')
```

```python
# 2. Initial Inspection
print("--- Dataset Info ---")
print(df.info())

# Summary statistics
print("\n--- Summary Statistics ---")
print(df.describe())
```

```python
# 3. Performing Exploratory Data Analysis and plotting the findings
# EDA on distributioin of selling price of cars

# Plot 1: Distribution of Selling Price
plt.figure(figsize=(10, 6))
sns.histplot(df['selling_price'], kde=True, color='teal')
plt.title('Distribution of Selling Price (Target Variable)')
plt.xlabel('Price in Lakhs')
plt.show()
```

```python
# Plot 2: Transmission Type Analysis
plt.figure(figsize=(10, 6))
sns.boxplot(x='transmission_type', y='selling_price', data=df, palette='Set2')
plt.title('Impact of Transmission Type on Price')
plt.show()
```

```python
# Plot 3: Correlation Heatmap
plt.figure(figsize=(12, 8))
numeric_cols = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='RdYlGn', fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.show()
```

```python
# 4. Insights
# Finding the average price per fuel type
fuel_analysis = df.groupby('fuel_type')['selling_price'].mean().sort_values(ascending=False)
print("Average Price by Fuel Type")
print(fuel_analysis)
```
